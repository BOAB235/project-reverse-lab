H   130061   A  /  HIJA01
 [1]    [2]   [3]      [4]


 [1] Material (H): This stands for High Flux (50% Nickel, 50% Iron alloy). If you open any competitor catalog (like Magnetics Inc.), they group their data by material first. You must look under the High Flux section.


 [2] Dimensions (130061): As established, this is the physical size. In a Changsung (CSC) catalog, you look for the 130061 page. In a Magnetics Inc. catalog, you look up 58061.


 [3] Permeability Code (A): This is the smoking gun that gives you your $A_L$ value. In the official CSC classification matrix, the letter A assigned to a High Flux core signifies a Permeability ($\mu$) of 125.


 [4] Tracking Code (HIJA01): This is the factory lot/batch number tracking the day, assembly line, and chemical blend sequence. It has zero impact on electrical performance and can be ignored during design.

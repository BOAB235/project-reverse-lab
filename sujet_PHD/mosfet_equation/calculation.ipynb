```python
# Id = Is (exp(Vd/(nVt)-1)
# Vt = kT/q
# n = ideality factor
import math
import sys
import numpy as np
import matplotlib.pyplot as plt
# V_t = ka * T
K_a= 8.617333262e-5
def Id(Vd, Is, T, n):
    T = T + 273.15 # convert to Kelvin
    Vt = K_a * T
    Id = Is * (np.exp(Vd/(n*Vt)) - 1)
    return Id
n=1.3 
Is = 7.9e-10
#Id(Vd, T, n)
Vd= np.arange(0.45, 1.4, 0.01)
id = Id(Vd=Vd, Is=Is, T=25, n=n)
plt.plot(Vd, id)
id = Id(Vd=Vd, Is=Is, T=125, n=n)
plt.plot(Vd, id)
plt.xlabel('Vd (V)')
plt.ylabel('Id (A)')
plt.title(f'Id vs Vd for T=25C and T=125C n={n},Is={Is}')
plt.legend(['T=25C', 'T=125C'])
plt.grid(which='both')
plt.yscale('log')
plt.show()
```

## Why the previous plot does NOT match the datasheet

**At Vsd = 1.4 V:** datasheet ~500 A, previous model ~1e8 A → off by ~10^6.

**Temperature is inverted:** datasheet shows 125 °C *left* of 25 °C (hotter = more current at same V). Previous model shows the opposite.

### Root causes
1. **`Is` was constant.** Physically `Is(T) ∝ T^3 · exp(-Eg / kT)`. For SiC, `Eg ≈ 3.26 eV`, so `Is` grows by many decades from 25 → 125 °C. This term dominates over the `1/(nVt)` in the exponent → without it, dI/dT has the wrong sign.
2. **No series resistance.** Above ~10 A the datasheet curve is linear (resistive), set by channel/drift `Rs`. Pure Shockley cannot bend.
3. **Si parameters used for a SiC device.** `Is = 7.9e-10`, `n = 1.3` are silicon values. SiC body diode needs much smaller `Is` and larger effective `n`.

### Correct model
$$V_{SD} = n\,V_T\,\ln\!\left(\frac{I}{I_S(T)}+1\right) + I\,R_s(T)$$

Sweep `I` (log-spaced), compute `V`. Fit `Is`, `n`, `Rs` to 2–3 points read off the datasheet at each temperature.

```python
# SiC body-diode model: Shockley + series resistance, with Is(T)
import numpy as np
import matplotlib.pyplot as plt

k_eV_per_K = 8.617333262e-5  # k_B / q  [V/K]
Eg_SiC = 3.26                # SiC bandgap [eV]

def Is_of_T(T_K, Is_ref=1e-25, T_ref=298.15, Eg=Eg_SiC):
    # Is(T) = Is_ref * (T/T_ref)^3 * exp(-Eg/k * (1/T - 1/T_ref))
    return Is_ref * (T_K / T_ref)**3 * np.exp(-(Eg / k_eV_per_K) * (1.0/T_K - 1.0/T_ref))

def Vsd_of_I(I, T_C, Is_ref, n, Rs):
    T_K = T_C + 273.15
    Vt = k_eV_per_K * T_K
    Is = Is_of_T(T_K, Is_ref=Is_ref)
    return n * Vt * np.log(I / Is + 1.0) + I * Rs

# Starting parameters tuned to roughly match Diagram 11 (IPQC60T022S7)
Is_ref = 1e-25      # at 25 C  -- tune to match knee
n       = 1.8
Rs_25   = 12e-3     # 12 mOhm
Rs_125  = 18e-3     # increases with T

I = np.logspace(-1, np.log10(100), 400)  # 0.1 A .. 700 A

V25  = Vsd_of_I(I, 25,  Is_ref, n, Rs_25)
V125 = Vsd_of_I(I, 125, Is_ref, n, Rs_125)

plt.figure()
plt.plot(V25,  I, label='T = 25 °C')
plt.plot(V125, I, label='T = 125 °C')
plt.yscale('log')
#plt.xlim(0, 1.4)
#plt.ylim(1e-1, 1e3)
plt.xlabel('Vsd (V)')
plt.ylabel('I_F (A)')
plt.title(f'SiC body diode  n={n}, Is_ref={Is_ref:.0e}, Rs(25)={Rs_25*1e3:.0f}mΩ, Rs(125)={Rs_125*1e3:.0f}mΩ')
plt.grid(which='both')
plt.minorticks_on()
plt.legend()
plt.show()
```

```python
# --- v2: empirical per-temperature parameters fitted to Diagram 11 ---
# Previous version used physical Is(T) with Eg=3.26 eV: shift was way too big and
# Is_ref=1e-25 put the knee at ~2.7 V (Si-like with n=1.8 puts knee much higher
# because n*Vt*ln(1/Is) blows up). Here we just fit Is and Rs per temperature.
import numpy as np
import matplotlib.pyplot as plt

k_eV_per_K = 8.617333262e-5  # V/K

def Vsd_model(I, T_C, Is, n, Rs):
    Vt = k_eV_per_K * (T_C + 273.15)
    return n * Vt * np.log(I / Is + 1.0) + I * Rs

# Targets read from Diagram 11 (IPQC60T022S7, Vgs=0):
#   25 C : ~1 A @ 0.80 V, ~10 A @ 0.95 V, ~100 A @ 1.15 V, ~500 A @ 1.40 V
#  125 C : ~1 A @ 0.65 V, ~10 A @ 0.85 V, ~100 A @ 1.10 V, ~500 A @ 1.40 V
# (curves cross around ~300-500 A because Rs(125) > Rs(25))

n      = 1.5
Is_25  = 3.0e-9     # sets knee at 25 C  (V(1A) ~ 0.80 V)
Is_125 = 5.0e-7     # ~170x higher  -> knee shift ~150 mV
Rs_25  = 6.0e-4     # 0.6 mOhm
Rs_125 = 1.0e-3     # 1.0 mOhm  (bigger Rs at hot)

I = np.logspace(-1, np.log10(700), 500)  # 0.1 .. 700 A

V25  = Vsd_model(I, 25,  Is_25,  n, Rs_25)
V125 = Vsd_model(I, 125, Is_125, n, Rs_125)

plt.figure(figsize=(7, 6))
plt.plot(V25,  I, label='T = 25 °C')
plt.plot(V125, I, label='T = 125 °C')
plt.yscale('log')
plt.xlim(0.0, 1.4)
plt.ylim(1e-1, 1e3)
plt.xlabel('Vsd (V)')
plt.ylabel('I_F (A)')
plt.title(f'SiC body diode  n={n}, Is(25)={Is_25:.0e}, Is(125)={Is_125:.0e}, '
          f'Rs(25)={Rs_25*1e3:.1f}mΩ, Rs(125)={Rs_125*1e3:.1f}mΩ')
plt.grid(which='both')
plt.minorticks_on()
plt.legend()
plt.show()
```

```python

# --- Robust fitting with separate Is per temperature ---
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator, LogLocator
from matplotlib.image import imread
from scipy.optimize import differential_evolution, least_squares

k_eV_per_K = 8.617333262e-5  # V/K

def Vsd_model(I, T_C, Is, n, Rs):
    Vt = k_eV_per_K * (T_C + 273.15)
    return n * Vt * np.log(I / Is + 1.0) + I * Rs

# --- Residuals in voltage space, weighted by 1/V to equalise log-scale importance ---
def residuals(params):
    Is_25, Is_125, n, Rs_25, Rs_125 = params
    I25  = df25.If.values
    I125 = df125.If.values
    V25_pred  = Vsd_model(I25,  25,  Is_25,  n, Rs_25)
    V125_pred = Vsd_model(I125, 125, Is_125, n, Rs_125)
    # relative voltage error  (avoids domination by high-current points)
    r25  = (V25_pred  - df25.vds.values)  / df25.vds.values
    r125 = (V125_pred - df125.vds.values) / df125.vds.values
    return np.concatenate([r25, r125])

def objective(params):
    return np.sum(residuals(params)**2)

# --- Stage 1: Global search with differential_evolution ---
bounds = [
    (1e-15, 1e-3),   # Is_25
    (1e-15, 1e-3),   # Is_125
    (0.5,   5.0),    # n
    (1e-5,  0.05),   # Rs_25
    (1e-5,  0.05),   # Rs_125
]
de_result = differential_evolution(
    objective, bounds,
    seed=42, maxiter=2000, tol=1e-14, polish=False,
    strategy='best1bin', mutation=(0.5, 1.5), recombination=0.9,
    popsize=30,
)
print(f"Differential Evolution  loss = {de_result.fun:.6e}")

# --- Stage 2: Local refinement with least_squares (Levenberg-Marquardt style) ---
ls_result = least_squares(
    residuals, de_result.x,
    bounds=([b[0] for b in bounds], [b[1] for b in bounds]),
    method='trf', ftol=1e-15, xtol=1e-15, gtol=1e-15, max_nfev=50000,
)
print(f"Least-squares refined   loss = {np.sum(ls_result.fun**2):.6e}")

Is_25, Is_125, n, Rs_25, Rs_125 = ls_result.x
print(f"\n--- Optimized parameters ---")
print(f"Is_25  = {Is_25:.4e} A")
print(f"Is_125 = {Is_125:.4e} A   (ratio Is_125/Is_25 = {Is_125/Is_25:.1f}x)")
print(f"n      = {n:.4f}")
print(f"Rs_25  = {Rs_25*1e3:.3f} mΩ")
print(f"Rs_125 = {Rs_125*1e3:.3f} mΩ")

# --- Evaluate model on datasheet points and on a dense sweep ---
I25  = df25.If.values
I125 = df125.If.values
V25_fit  = Vsd_model(I25,  25,  Is_25,  n, Rs_25)
V125_fit = Vsd_model(I125, 125, Is_125, n, Rs_125)

I_dense = np.logspace(-2, np.log10(700), 600)
V25_dense  = Vsd_model(I_dense, 25,  Is_25,  n, Rs_25)
V125_dense = Vsd_model(I_dense, 125, Is_125, n, Rs_125)

# --- Per-point errors ---
print(f"\n--- 25 °C fit errors ---")
for v_ds, v_fit, i_f in zip(df25.vds.values, V25_fit, I25):
    print(f"  I={i_f:8.2f} A   Vds_data={v_ds:.4f}   Vds_model={v_fit:.4f}   err={v_fit-v_ds:+.4f} V")
print(f"\n--- 125 °C fit errors ---")
for v_ds, v_fit, i_f in zip(df125.vds.values, V125_fit, I125):
    print(f"  I={i_f:8.2f} A   Vds_data={v_ds:.4f}   Vds_model={v_fit:.4f}   err={v_fit-v_ds:+.4f} V")

# --- Plot ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 6))

# Subplot 1: fit vs datasheet
ax1.plot(V25_dense,  I_dense, c="blue",  label='Model 25 °C')
ax1.plot(V125_dense, I_dense, c="red",   label='Model 125 °C')
ax1.plot(df25.vds,   I25,  'o', c="blue", markersize=5, label='Data 25 °C', linestyle='--')
ax1.plot(df125.vds,  I125, 's', c="red",  markersize=5, label='Data 125 °C', linestyle='--')
ax1.set_yscale('log')
ax1.set_xlim(0.0, 1.6)
ax1.set_ylim(1e-1, 1e3)
ax1.set_xlabel('Vsd (V)')
ax1.set_ylabel('I_F (A)')
ax1.set_title(f'Optimized: n={n:.3f}, Is25={Is_25:.1e}, Is125={Is_125:.1e}\n'
              f'Rs25={Rs_25*1e3:.2f}mΩ, Rs125={Rs_125*1e3:.2f}mΩ')
ax1.xaxis.set_major_locator(MultipleLocator(0.2))
ax1.xaxis.set_minor_locator(MultipleLocator(0.1))
ax1.yaxis.set_minor_locator(LogLocator(base=10, subs=np.arange(2, 10) * 0.1, numticks=100))
ax1.grid(which='major', linewidth=0.8)
ax1.grid(which='minor', linewidth=0.3)
ax1.legend(fontsize=9)

# Subplot 2: datasheet image
img = imread('datasheet1.png')
ax2.imshow(img)
ax2.set_title('Datasheet: Diagram 11')
ax2.axis('off')

plt.tight_layout()
plt.show()
```

### Temperature dependence of $I_S$

$$I_S(T) = I_{S,\text{ref}} \cdot \left(\frac{T}{T_\text{ref}}\right)^3 \cdot \exp\!\left(-\frac{E_g}{k}\left(\frac{1}{T} - \frac{1}{T_\text{ref}}\right)\right)$$

where:
- $I_{S,\text{ref}}$ is the saturation current at reference temperature $T_\text{ref}$
- $E_g$ is the bandgap energy (SiC ≈ 3.26 eV)
- $k = 8.617 \times 10^{-5}$ eV/K (Boltzmann constant / elementary charge)
- $T$ in Kelvin

```python
#Is_25  = 3.7760e-10 A
#Is_125 = 3.5662e-06 A   (ratio Is_125/Is_25 = 9444.6x)
Is_125/Is_25
```

```python

# From the two fitted Is values, extract Eg and Is_ref using the physical model:
#   Is(T) = Is_ref * (T/T_ref)^3 * exp(-Eg/k * (1/T - 1/T_ref))
# With T_ref = 298.15 K (25°C), we know Is_25 = Is_ref.
# From Is_125 / Is_25 we solve for Eg.

import numpy as np

k = 8.617333262e-5  # eV/K

T1 = 25 + 273.15    # 298.15 K
T2 = 125 + 273.15   # 398.15 K

# Use T_ref = T1, so Is_ref = Is_25
Is_ref = Is_25
T_ref  = T1

ratio = Is_125 / Is_25
print(f"Is_125 / Is_25 = {ratio:.2f}")

# ratio = (T2/T1)^3 * exp(-Eg/k * (1/T2 - 1/T1))
# => ln(ratio) = 3*ln(T2/T1) - Eg/k * (1/T2 - 1/T1)
# => Eg = (3*ln(T2/T1) - ln(ratio)) * k / (1/T2 - 1/T1)
Eg = (3 * np.log(T2 / T1) - np.log(ratio)) * k / (1/T2 - 1/T1)

print(f"\n--- Physical parameters ---")
print(f"Is_ref = {Is_ref:.4e} A   (at T_ref = {T_ref:.2f} K = {T_ref-273.15:.0f} °C)")
print(f"Eg     = {Eg:.4f} eV")
print(f"k      = {k:.6e} eV/K")

# Verify: recompute Is at both temperatures
def Is_of_T(T_K):
    return Is_ref * (T_K / T_ref)**3 * np.exp(-(Eg / k) * (1/T_K - 1/T_ref))

Is_25_check  = Is_of_T(T1)
Is_125_check = Is_of_T(T2)

print(f"\n--- Verification ---")
print(f"Is(25°C)  = {Is_25_check:.4e} A   (fitted: {Is_25:.4e} A)")
print(f"Is(125°C) = {Is_125_check:.4e} A   (fitted: {Is_125:.4e} A)")
print(f"Match: {np.isclose(Is_25_check, Is_25) and np.isclose(Is_125_check, Is_125)}")

print(f"\n--- Note ---")
print(f"SiC theoretical Eg ≈ 3.26 eV")
print(f"Fitted Eg = {Eg:.4f} eV  →  {'close to SiC' if 2.5 < Eg < 4.0 else 'differs from SiC'}")
```

```python

# --- Two-diode model (standard pn-junction physics, NOT arbitrary) ---
# Every real diode has two current mechanisms:
#   1) Diffusion current:     I₁ = Is1 · exp(Vj / Vt)       → n₁ = 1 (FIXED)
#   2) Recombination current: I₂ = Is2 · exp(Vj / (2·Vt))   → n₂ = 2 (FIXED)
# n₁ and n₂ are NOT fitted — they come from semiconductor physics.
# This is the standard SPICE diode model.
#
# Total: I = Is1·exp(Vj/Vt) + Is2·exp(Vj/(2Vt))
# Quadratic in u = exp(Vj/(2Vt)):
#   Is1·u² + Is2·u - I = 0
#   u = (-Is2 + √(Is2² + 4·Is1·I)) / (2·Is1)
#   Vj = 2·Vt·ln(u),  Vds = Vj + I·Rs
#
# 6 params fitted: Is1_25, Is1_125, Is2_25, Is2_125, Rs_25, Rs_125
# 0 free ideality factors.

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator, LogLocator
from matplotlib.image import imread
from scipy.optimize import differential_evolution, least_squares

k_eV_per_K = 8.617333262e-5

def Vsd_two_diode(I, T_C, Is1, Is2, Rs):
    Vt = k_eV_per_K * (T_C + 273.15)
    disc = Is2**2 + 4.0 * Is1 * I
    u = (-Is2 + np.sqrt(disc)) / (2.0 * Is1)
    Vj = 2.0 * Vt * np.log(u)
    return Vj + I * Rs

def residuals_2d(params):
    Is1_25, Is1_125, Is2_25, Is2_125, Rs_25, Rs_125 = params
    V25_p  = Vsd_two_diode(df25.If.values,  25,  Is1_25,  Is2_25,  Rs_25)
    V125_p = Vsd_two_diode(df125.If.values, 125, Is1_125, Is2_125, Rs_125)
    r25  = (V25_p  - df25.vds.values)  / df25.vds.values
    r125 = (V125_p - df125.vds.values) / df125.vds.values
    return np.concatenate([r25, r125])

def objective_2d(params):
    return np.sum(residuals_2d(params)**2)

bounds_2d = [
    (1e-40, 1e-1),   # Is1_25  (diffusion, n=1)
    (1e-40, 1e-1),   # Is1_125
    (1e-40, 1e-1),   # Is2_25  (recombination, n=2)
    (1e-40, 1e-1),   # Is2_125
    (1e-5,  0.05),   # Rs_25
    (1e-5,  0.05),   # Rs_125
]

de_2d = differential_evolution(
    objective_2d, bounds_2d,
    seed=42, maxiter=2000, tol=1e-14, polish=False,
    strategy='best1bin', mutation=(0.5, 1.5), recombination=0.9, popsize=30,
)
print(f"DE loss = {de_2d.fun:.6e}")

ls_2d = least_squares(
    residuals_2d, de_2d.x,
    bounds=([b[0] for b in bounds_2d], [b[1] for b in bounds_2d]),
    method='trf', ftol=1e-15, xtol=1e-15, gtol=1e-15, max_nfev=50000,
)
loss_2d = np.sum(ls_2d.fun**2)

Is1_25f, Is1_125f, Is2_25f, Is2_125f, Rs_25_2d, Rs_125_2d = ls_2d.x

# Compare with cell 29 (single Shockley)
old_loss = np.sum(residuals(ls_result.x)**2)
print(f"\nShockley 1-diode (5p, shared n) loss = {old_loss:.6e}")
print(f"Two-diode (6p, n₁=1 n₂=2 fixed) loss = {loss_2d:.6e}")
print(f"Improvement: {(1 - loss_2d/old_loss)*100:.1f}%")

print(f"\n--- Parameters ---")
print(f"Is1_25  = {Is1_25f:.4e} A   (diffusion, n=1)")
print(f"Is1_125 = {Is1_125f:.4e} A")
print(f"Is2_25  = {Is2_25f:.4e} A   (recombination, n=2)")
print(f"Is2_125 = {Is2_125f:.4e} A")
print(f"Rs_25   = {Rs_25_2d*1e3:.3f} mΩ")
print(f"Rs_125  = {Rs_125_2d*1e3:.3f} mΩ")

# Per-point errors
I25 = df25.If.values; I125 = df125.If.values
V25_2d = Vsd_two_diode(I25, 25, Is1_25f, Is2_25f, Rs_25_2d)
V125_2d = Vsd_two_diode(I125, 125, Is1_125f, Is2_125f, Rs_125_2d)
print(f"\n--- 25 °C errors ---")
for vd, vm, ii in zip(df25.vds.values, V25_2d, I25):
    print(f"  I={ii:7.1f} A  data={vd:.4f}  model={vm:.4f}  err={vm-vd:+.4f} V")
print(f"\n--- 125 °C errors ---")
for vd, vm, ii in zip(df125.vds.values, V125_2d, I125):
    print(f"  I={ii:7.1f} A  data={vd:.4f}  model={vm:.4f}  err={vm-vd:+.4f} V")

# Plot
I_d = np.logspace(-2, np.log10(700), 600)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 6))
ax1.plot(Vsd_two_diode(I_d, 25, Is1_25f, Is2_25f, Rs_25_2d), I_d, c="blue", label='Model 25 °C')
ax1.plot(Vsd_two_diode(I_d, 125, Is1_125f, Is2_125f, Rs_125_2d), I_d, c="red", label='Model 125 °C')
ax1.plot(df25.vds, I25, 'o', c="blue", ms=5, label='Data 25 °C')
ax1.plot(df125.vds, I125, 's', c="red", ms=5, label='Data 125 °C')
ax1.set_yscale('log'); ax1.set_xlim(0, 1.6); ax1.set_ylim(1e-1, 1e3)
ax1.set_xlabel('Vsd (V)'); ax1.set_ylabel('I_F (A)')
ax1.set_title('Two-diode model (n₁=1, n₂=2 — fixed by physics)\n'
              f'Rs25={Rs_25_2d*1e3:.2f}mΩ, Rs125={Rs_125_2d*1e3:.2f}mΩ')
ax1.xaxis.set_major_locator(MultipleLocator(0.2)); ax1.xaxis.set_minor_locator(MultipleLocator(0.1))
ax1.yaxis.set_minor_locator(LogLocator(base=10, subs=np.arange(2,10)*0.1, numticks=100))
ax1.grid(which='major', lw=0.8); ax1.grid(which='minor', lw=0.3); ax1.legend(fontsize=9)
img = imread('datasheet1.png'); ax2.imshow(img); ax2.set_title('Datasheet'); ax2.axis('off')
plt.tight_layout(); plt.show()
```

# ZVS simulation
Half-bridge: 2 MOSFET Coss capacitors, constant inductor current $I_L$.  
High-side turns OFF → $I_L$ discharges $C_{high}$ and charges $C_{low}$.  
When $V_{mid} < 0$ → low-side body diode conducts: $V_{mid} = -V_{SD}(I_L, T)$

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# --- Circuit parameters ---
Vbus = 400          # DC bus [V]
I_L  = 10           # constant inductor current [A] (ZVS current)
Coss = 200e-12      # output capacitance per MOSFET [F]

# --- Initial conditions: high-side was ON ---
# Vc_high(0) = 0,  Vc_low(0) = Vbus = 400V
# V_mid = Vbus - Vc_high = Vc_low  (when capacitive)

Vsd_clamp = vds_of_I(I=I_L, T_C=25)
print(f"Body diode Vsd({I_L}A, 25°C) = {Vsd_clamp:.4f} V")
print(f"When diode ON: V_mid = -{Vsd_clamp:.4f} V")

# Phase 1 (capacitive): dV_mid/dt = -I_L / (2*Coss)
#   V_mid(t) = Vbus - I_L*t / (2*Coss)
# Phase 2 (diode clamp): V_mid = -Vsd

def deriv(t, y):
    V_mid = y[0]
    if V_mid > -Vsd_clamp:
        return [-I_L / (2 * Coss)]
    return [0.0]

# Event: V_mid crosses -Vsd_clamp
evt = lambda t, y: y[0] + Vsd_clamp
evt.terminal = True
evt.direction = -1

t_end = 20e-9
sol = solve_ivp(deriv, (0, t_end), [Vbus], events=evt, max_step=0.05e-9)

t_diode = sol.t_events[0][0] if sol.t_events[0].size else t_end
print(f"Diode turns ON at t = {t_diode*1e9:.2f} ns")
print(f"t_ZVS = 2·Coss·Vbus/I_L = {2*Coss*Vbus/I_L*1e9:.2f} ns")

# Build full waveform
t = np.linspace(0, t_end, 2000)
V_mid = np.where(t < t_diode, Vbus - I_L * t / (2 * Coss), -Vsd_clamp)

plt.figure(figsize=(9, 4))
plt.plot(t * 1e9, V_mid, 'b', lw=2)
plt.axhline(0, color='k', lw=0.5)
plt.axhline(-Vsd_clamp, color='r', ls='--', label=f'diode clamp = {-Vsd_clamp:.3f} V')
plt.xlabel('t [ns]'); plt.ylabel('V_middle [V]')
plt.title(f'ZVS transition: Vbus={Vbus}V, I_L={I_L}A, Coss={Coss*1e12:.0f}pF')
plt.legend(); plt.grid(True); plt.tight_layout(); plt.show()
```

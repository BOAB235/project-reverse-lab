```python
import html2text

# Input and output file paths
html_file = "ESP32_PWM_DAC_Function_Generator_V1.html"
md_file = "readme__.md"

# Read HTML content
with open(html_file, "r", encoding="utf-8") as f:
    html_content = f.read()

# Convert HTML to Markdown
markdown_converter = html2text.HTML2Text()
markdown_converter.ignore_links = False
markdown_converter.ignore_images = False
markdown_converter.body_width = 0  # No line wrapping
markdown_content = markdown_converter.handle(html_content)

# Save Markdown
with open(md_file, "w", encoding="utf-8") as f:
    f.write(markdown_content)

print(f"Converted '{html_file}' to Markdown and saved as '{md_file}'.")
```
